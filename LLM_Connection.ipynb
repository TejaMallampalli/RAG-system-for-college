# *LLM CONNECTION*

## *LLM Connection for Answer Generation*

### *The retrieved context and user question are sent to Ollama LLM like llama3.2. The LLM reads the context and generates a proper human-like answer. It answers only from the given context, not from its own memory.*

```python
from langchain_ollama import ChatOllama
```

```python
llm = ChatOllama(
    model = "ollama3.2",
    temperature = 0
)
```

```python
response = llm.invoke("What is the CGPA for Second Class?")

print(response.content)
```
