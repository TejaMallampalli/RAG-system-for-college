## *Retrieval / Similarity Search*

### *When a user asks a question, we convert the question into a vector and search for the most similar chunks in ChromaDB. This is called similarity search. It returns the most relevant context for the question.*

```python
question = "What is the CGPA for First Class?"
```

```python
q_vector = embeddings.embed_query(question)
```

```python
result = collection.query(
    query_embeddings = [q_vector],
    n_results = 1
)

print(result['documents'][0])
```
